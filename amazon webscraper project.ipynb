```python
import datetime
import csv
import os
import pandas as pd
from bs4 import BeautifulSoup


file_name = r"E:\Practice\Amazonscrapingdata.csv"
html_file = r"E:\Practice\.ipynb_checkpoints\.ipynb_checkpoints\amazon_page.txt.txt"   

def check_price():
    with open(html_file, "r", encoding="utf-8") as f:
        html = f.read()

    soup = BeautifulSoup(html, "html.parser")

    # Extract title
    title_tag = soup.find(id="productTitle")
    title = title_tag.get_text().strip() if title_tag else "Title not found"

    # Extract price
    symbol = soup.select_one("span.a-price-symbol")
    whole = soup.select_one("span.a-price-whole")
    fraction = soup.select_one("span.a-price-fraction")

    if symbol and whole and fraction:
        price = (symbol.text + whole.text + fraction.text).strip()
        price = price.replace("\n", "").replace(" ", "")
    else:
        price = "Price not found"

    price = price.strip()[1:] if price != "Price not found" else price

    today = datetime.date.today()

    header = ["title", "price", "date"]
    data = [title, price, today]

    file_exists = os.path.isfile(file_name)

    with open(file_name, "a", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        if not file_exists:
            writer.writerow(header)
        writer.writerow(data)

    return data

# Test run once
row = check_price()
print("Scraped row:", row)

# Show CSV
df = pd.read_csv(file_name)
print(df)
```
